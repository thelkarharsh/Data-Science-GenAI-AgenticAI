# 19.write a program that check if the input number is a prime number.

```python
num = int(input("Enter a num to check prime or Not: "))
if num <= 1:
   print("Not Prime")
else:
    for i in range(2, num):
      if num % i == 0:
         print("Not Prime")
         break
    else:
       print("Prime")
```
